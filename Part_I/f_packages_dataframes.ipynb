## 🟢 Information: Creating a dataframe

A dataframe is a table of rows and columns.

## 🟢 Information: Filtering rows with Boolean conditions

In Notebook c, comparing two values produced a Boolean (`True` or `False`).

In pandas, we can compare a whole **column** with a value to produce a whole column of Booleans:
one for each row.

````python
musicians["birth_year"] < 1980
````

We can use this column of `True` / `False` values as a filter.

If you put it inside square brackets after the dataframe name, pandas keeps only the rows marked `True`.

Read the next cell as: "from the musicians dataframe, keep the rows where `birth_year` is below 1980".

````python
musicians[musicians["birth_year"] < 1980]
````

This works with text too.
As in Notebook c, remember to use `==` (double) to compare:

````python
musicians[musicians["borough"] == "Lambeth"]
````

We can combine conditions with two special characters for our Boolean operators:
- `&` is for "and" and
- `|` stands for "or".

Each condition (either side of the operator) must be in its own round brackets.

````python
musicians[(musicians["birth_year"] < 1980) & (musicians["borough"] != "Lambeth")]
````

````python
musicians[(musicians["borough"] == "Lambeth") | (musicians["borough"] == "Harrow")]
````

📝 Note: In Notebook c we combined conditions with the words `and` and `or`.
Those only work on a single `True` / `False` value.
For a whole column of them we need the symbols `&` and `|`, which work row by row.
Using `and` / `or` here gives an error (see Common Errors below).

(There is also a symbol `~` for "not", but `!=` and the other comparisons cover most needs, so we won't use it here.)

There's also a shortcut for "is this value one of these?".
In Notebook b, we checked `"Brent" in borough_population`.
The column version is `.isin(...)`, which takes a list:

````python
musicians[musicians["borough"].isin(["Lambeth", "Harrow", "Barnet"])]
````

📝 Note: square brackets after a dataframe do three different jobs, depending on what goes inside them:
- a single column name (`df["a"]`) gives that column;
- a *list* of column names (`df[["a", "b"]]`) gives those columns;
- a column of `True` / `False` values (`df[df["a"] > 5]`) gives the rows marked `True`.

## 🟢 Information: Using your own functions with `.apply`

Now let's restore a version of our data about London venues and turn that into a dataframe.

(As before, prices and visitor numbers are illustrative, not exact.)

````python
venues = pd.DataFrame(
    {
        "venue": [
            "Tate Modern",
            "London Eye",
            "Shakespeare's Globe",
            "British Museum Guided Tour",
            "Cutty Sark",
            "Science Museum",
        ],
        "type": ["gallery", "attraction", "theatre", "tour", "museum", "museum"],
        "price": [0, 29, 25, 90, 18, 0],
        "visitors_m": [5.9, 3.5, 0.4, 0.1, 0.3, 3.0],
    }
)

venues
````

And let's recreate a version of the `price_tier` function we made in Notebook e.

````python
def price_tier(price):
    if price == 0:
        return "Free"
    elif price < 20:
        return "Budget"
    elif price < 40:
        return "Standard"
    else:
        return "Premium"
````

To run this function on every value in a column, use `.apply`.

This method calls the named function once for each value, and collects the results in a new column.

````python
venues["tier"] = venues["price"].apply(price_tier)
venues
````

📝 Note: look closely at `.apply(price_tier)`.
The function name has **no round brackets**.

In Notebook e, forgetting the brackets was a mistake: Python just referred to the function without running it.
Here that's exactly what we want.

We *hand over* the function itself to pandas.
We simply provide the name of the function,
and then `.apply` does the calling for us, once per value.
If we wrote `.apply(price_tier())`, Python would try to run the function immediately with no input, and fail.

This has the same result as looping over the column (Notebook d) and calling the function each time,
but it's shorter and pandas does the looping for us.

`sort_values` sorts the rows by the values in a column.

As with `sorted` in Notebook b,
- the default is ascending order (smallest first).
- use `ascending=False` to get descending order (largest first).

````python
venues.sort_values("visitors_m", ascending=False)
````

As with `sorted`, this returns a **new** dataframe: the original `venues` is unchanged.

````python
venues
````

(If you want to keep a sorted version, store it in a variable: `venues_sorted = venues.sort_values(...)`.)

Methods can be combined.

In Python, we say that the methods are "chained",
and we work left to right.

In the following cell take the venues data then:
1. First, sort (`sort_values`)
2. Second, keep only the top three rows (with `.head(3)`):

````python
venues.sort_values("visitors_m", ascending=False).head(3)
````

Summary functions, as in Notebook b (`max`, `sum`, etc.), work on a column:

````python
venues["price"].mean()
````

````python
venues["visitors_m"].sum().round(1)
````

📝 Note: we added `.round(1)` to tidy the answer to one decimal place.
Without it, the computer may show something like `13.200000000000001`:
computers store decimals as very slightly inexact approximations, so tiny errors like this can appear.
Nothing is wrong: just round the result when you display it.

## 🟢 Information: Loops versus pandas

In Notebook d we counted how many parks were larger than 100 hectares, using the accumulator pattern.
Here's the same pattern applied to the `price` column: how many venues cost more than £20?

````python
count = 0

for price in venues["price"]:
    if price > 20:
        count = count + 1

count
````

That works: a column can be looped over just like a list.
But pandas has a much shorter way.
A comparison gives a column of Booleans, and `True` counts as `1` while `False` counts as `0`.
So summing the Booleans counts the `True`s:

````python
(venues["price"] > 20).sum()
````

📝 Note: in pandas, a loop is usually longer and slower than the built-in alternative.
Write a loop if you need to, but check first whether a pandas method already does the job.
This advice applies to coding tasks well beyond pandas:
do not "reinvent the wheel"!

## ⚠️ Common Errors with Packages and Dataframes

📝 Note: the code cells in this section are **supposed** to go wrong.

We expect an error message, or a strange result: that's the point!

Running them does no harm, and seeing each mistake once makes it easier to recognise next time.

### Packages

**1. A typo in the package name**

Python can't find a package that doesn't exist (or that isn't installed): `ModuleNotFoundError`.

````python
import panda
````

Fix: check the spelling (it's `pandas` with an `s`).
If the spelling is right and the error persists, the package probably isn't installed yet:
run `%pip install pandas` and then try again.
If you still see the error after installing, restart the kernel.

**2. Using the full name after importing it under a nickname**

We wrote `import pandas as pd`, so from then on the package is called `pd`, and `pandas` on its own is not defined: `NameError`.

````python
pandas.DataFrame({"a": [1, 2]})
````

Fix: use `pd`.

Similarly, note that if you restart the kernel, all imports and variables are forgotten.
If you see `NameError: name 'pd' is not defined`, re-run `import pandas as pd`

**3. pip refuses to install: `externally-managed-environment`**

Some computers protect their built-in ("system") Python and refuse to install packages into it.
`%pip install` then fails with a message containing `externally-managed-environment`.
Nothing is wrong with the package: you're using the wrong Python.

Fix: use the Python environment that you set up for this course
(e.g., the one built from the requirements file),
and start Jupyter from inside it.

Don't work around the message by forcing the install:
that risks breaking your computer's own tools.

### Dataframes

**4. Columns of different lengths**

Every list in the dict must have the same length: `ValueError`.

````python
pd.DataFrame({"a": [1, 2, 3], "b": [1, 2]})
````

Fix: count the values in each list.

**5. Wrong column name**

Column names must match exactly, including capital letters: `KeyError` (as with dicts in Notebook b).

````python
musicians["Name"]
````

Fix: check `musicians.columns` for the exact spelling.

**6. Brackets on an attribute**

`.shape` is an attribute (information), not a method (an action), so it takes no round brackets.
Adding them tries to "call" the result (a pair of numbers, like `(5, 4)`): `TypeError`.

````python
musicians.shape()
````

Fix: remove the brackets.
(And, if you forget them on a *method* like `.head`,
you get the method object printed back at you, as in Notebook e.)

**7. Using `and` instead of `&`**

`and` and `or` work on single `True` / `False` values, not on whole columns: `ValueError`.

````python
musicians[(musicians["birth_year"] < 1980) and (musicians["borough"] == "Lambeth")]
````

Fix: use `&` (or `|`), with each condition in round brackets.

Forgetting the round brackets is a related mistake:
`&` is applied *before* the comparisons, which gives a different error (a `TypeError`).

````python
musicians[musicians["birth_year"] < 1980 & musicians["borough"] == "Lambeth"]
````

Fix: put each condition in round brackets.

**8. Round brackets instead of square brackets with `iloc`**

`iloc` is looked up with square brackets.
With round brackets there is usually no helpful error message:
you just get a strange-looking object back instead of your row.

````python
musicians.iloc(0)
````

Fix: use square brackets: `musicians.iloc[0]`.

**9. Calling the function inside `.apply`**

`.apply` needs the function itself, with no round brackets.
Adding them runs the function straight away, with no input, thus raising a `TypeError`.

````python
venues["price"].apply(price_tier())
````

Fix: `venues["price"].apply(price_tier)`.

**10. Forgetting to store the result**

There's no error here, which makes this one easy to miss.

`sort_values` returns a new dataframe, so `venues` itself stays as it was, as with `sorted` in Notebook b.

````python
venues.sort_values("price")
venues["price"]
````

The prices are still in the original order.

Fix: store the result: `venues_sorted = venues.sort_values("price")`,
or overwrite the original with `venues = venues.sort_values("price")`.
