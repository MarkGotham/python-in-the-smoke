## 🟢 Information: Why functions?

We've already used lots of functions,
previously calling them "commands":
`print`, `len`, `round`, `type`, `sorted`, `sum`, `max`, `min`.

So far we've used only functions that Python provides ready-made ('built-in').

A **function** is a named, reusable block of code.
It typically:
- takes some input (sometimes none),
- does something with it,
- and gives something back (again, usually, not always).

So far we've only *called* functions that someone else wrote.
In this notebook, we'll *write our own*.

Why bother?
Two of the main reasons are:
- **Don't repeat yourself!**
If you find yourself copying and pasting the same few lines of code,
changing only one or two values, that's a sign you want a function instead.
This is a central principle of coding and even goes by the acronym "DRY".
- **Give a name to what your code does.**
A well-named function makes the rest of your code easier to read,
because then you don't need to reread the details to know what that code is for.

## 🟢 Information: Defining a function

Here's a very simple function.

````python
def greet():
    print("Hello, welcome to the Tate Modern!")
````

Notice:

1. The `def` keyword starts the definition, followed on the same line by:
    - A space (don't forget that!)
    - The function's **name** (`greet`),
    - Round brackets `()`,
    - A colon `:`.
2. After the `def` line comes one or more lines below, all of which are **indented**.
This is just the same as with `if` (Notebook c) and `for` (Notebook d).
3. Running the cell above doesn't print anything.
It just teaches Python the recipe called `greet`; it doesn't run it yet.
4. Running the cell below calls the function, runs it, and sure enough prints something.

````python
greet()
````

Now it runs.

Notice the round brackets `()` when we **call** the function.

That's exactly the bracket syntax we've used when calling `print()`, `len()`, and so on.

## 🟢 Information: Parameters and arguments

The `greet` and `borough_fact` functions always do exactly the same thing, however many times we call them.
That's fine, but kind of limited.
Usually we want a function to work with different data each time depending on changing input.
That's what **parameters** are for.

A parameter is a variable listed inside the function's brackets.
When we call the function, we give it a value (called an **argument**) to fill that variable.

````python
def greet_visitor(name):
    print(f"Hello, " + str(name), "and welcome to the Tate Modern!")
````

````python
greet_visitor("Amara")
````

````python
greet_visitor("Tom")
````

`name` is the **parameter**: a placeholder used inside the function.

`"Amara"` and `"Tom"` are the **arguments**: the actual values supplied when we call the function.

Each call can supply a different argument, and the function reuses the same code with that new value substituted in.

## 🟢 Information: f-strings for mixing text and numbers

📝 If you did the "⭐ Optional Exercise" at the end of notebook a, this is a reminder;
if not, let's get you up to speed.

So far, to mix text and numbers we've had to convert the number to a string first with `str()`.

Python has a shortcut called an **f-string**:
put an `f` right before the opening quote mark,
and put any variable or number inside `{}`.

With this function, no `str()` or `+` are needed.

For example:

````python
exhibition = "Tate Modern"
visitors = 5900000
print(f"{exhibition} welcomes about {visitors} visitors every year.")
````

With this, we can make functions like `greet_visitor` and `describe_venue` that bit neater:

````python
def greet_visitor(name):
    print(f"Hello, {name}, welcome to the Tate Modern!")
````

````python
greet_visitor("Amara")
greet_visitor("Tom")
````

````python
def describe_venue(venue):
    print(f"{venue} is a great place to visit.")
````

````python
describe_venue("Tate Modern")
describe_venue("The Globe Theatre")
````

From now on, we'll use f-strings for cases like this.

## 🟢 Information: Multiple parameters

A function can take more than one parameter, separated by commas.

````python
def venue_summary(venue, price):
    print(f"{venue} costs £{price} to visit.")
````

````python
venue_summary("London Eye", 29)
````

📝 Note:
- Arguments are matched to parameters **in order**.
- Here, `"London Eye"` fills `venue` and `29` fills `price`because that's the order they're listed in `def venue_summary(venue, price):`.

Swap the order of the arguments in the cell above and re-run it, to see what happens when this goes wrong.

````python
venue_summary(29, "London Eye")
````

... Not so good is it?

## 🟢 Information: Returning a value

So far, our functions have **printed** something, but haven't given a value back for later use.

This is an important difference:
- `print` *displays* a value once.
- `return` *hands the value back* to whoever called the function, so it can be stored in a variable or used in a further calculation.

````python
def double(number):
    return number * 2
````

````python
double(6)
````

So far, this looks similar to what `print` would give us, but it's not the same thing.

A returned value can be **stored** as a variable,
exactly like the result of any built-in function,
or anything newly assigned.

````python
result = double(6)
result
````

📝 Note: if a function has no `return` line,
Python gives back a special value called `None`,
even if the function has printed something along the way.
This is a common source of confusion. See the Common Errors below.

## 🟢 Information: Functions can use everything you've learned so far

A function is just a named block of code
It can contain variables (see Notebook a),
conditionals (see Notebook c), and
loops (see Notebook d),
exactly like any other code.

Here, we adapt a price-tier `if` / `elif` / `else` chain from Notebook c into a function.

````python
def price_tier(price):
    if price < 0:
        return "Paid to enter!"
    if price == 0:
        return "Free"
    elif price < 20:
        return "Budget"
    elif price < 40:
        return "Standard"
    else:
        return "Premium"
````

````python
price_tier(0)
````

````python
price_tier(29)
````

Because `price_tier` is now a reusable function,
we can use it inside a loop (Notebook d),
applying it to every venue in a dictionary at once:

````python
venue_prices = {"Tate Modern": 0, "London Eye": 29, "Shakespeare's Globe": 25}

for venue, price in venue_prices.items():
    print(venue, "is", price_tier(price))
````

This does the same job as the loop-plus-`elif` from Notebook d,
but now the price-tier logic lives in **one place** (`price_tier`).

This can now be reused anywhere:
e.g., on a single price, inside this loop, or inside a completely different loop over a different dictionary.

Define a function called `borough_size` that takes one parameter, `population`, and **returns**:

### ✅ Solution: Functions, conditionals, and loops together

````python
borough_population = {
    "Camden": 210000,
    "Islington": 245000,
    "Hackney": 281000,
    "Brent": 350000,
}


def borough_size(population):
    if population < 220000:
        return "Small"
    elif population <= 260000:
        return "Medium"
    else:
        return "Large"
````

## ⚠️ Common Errors

Three errors to look out for, and how to fix them.

**1. Calling a function without parentheses**

Writing a function's name **without** `()` doesn't run it,
it simply refers to the function itself.

````python
borough_fact
````

This is a 'common error' in our sense,
but it doesn't actually trigger any Python errors:
Python simply shows you the function object itself, rather than running it.

That is one of the reasons why this is an easy error to make,
and to miss when reviewing your code.

Fix: add the parentheses to actually call it.

````python
borough_fact()
````

**2. Wrong number of arguments**

Every parameter in the function's definition needs a matching argument when you call it
(unless it has a default value as shown in the optional exercise below).

````python
describe_venue()
````

This gives a `TypeError`, telling you an argument is missing.

Note that it also tells you which argument is missing.
Here' the error message is `TypeError: describe_venue() missing 1 required positional argument: 'venue'`
indicating that the `venue` is missing.

Fix: supply an argument for every parameter.

````python
describe_venue("Tate Modern")
````

**3. Forgetting to `return`**

A function that `print`s a result looks like it's "given you" a value, but if you try to store that value in a variable, you won't get what you expect.

````python
def triple(number):
    print(number * 3)
````

````python
result = triple(4)
result
````

`triple(4)` prints `12`, so it looks like it worked,
but `result` is actually `None`, because the function never used `return`.

Remember: `print` only *displays* a value; it doesn't hand it back.

Fix: use `return` instead of (or as well as) `print`, whenever you need the function's result for later use.
