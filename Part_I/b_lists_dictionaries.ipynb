## 🟢 Information: Indices

Each item in a list has a position, called an **index** (plural: 'indices').

Important! Indices start at `0`, not `1`.

It's a bit like riding the London underground ;)

Let's get on the circle line, heading clockwise, starting at our local stop: Temple.

````python
circle_line_stops = [
    "Temple",
    "Embankment",
    "Westminster",
    "St James's Park",
    "Victoria",
    "Sloane Square",
    "South Kensington",
    "Gloucester Road",
    "High Street Kensington",
    "Notting Hill Gate",
    "Bayswater",
    "Paddington",
    "Edgware Road",
    "Baker Street",
    "Great Portland Street",
    "Euston Square",
    "King's Cross St Pancras",
    "Farringdon",
    "Barbican",
    "Moorgate",
    "Liverpool Street",
    "Aldgate",
    "Tower Hill",
    "Monument",
    "Cannon Street",
    "Mansion House",
    "Blackfriars",
]
````

The first station we encounter (remember: index 0 not 1) is Temple:

````python
circle_line_stops[0]
````

The next station is ...

````python
circle_line_stops[1]
````

Negative indices count backwards from the end of the list.

`-1` is always the last item, whatever the length of the list.

Here on the circle line, that also means the first step in the opposite direction.

On other lines, it would mean the last station (terminus).

````python
circle_line_stops[-1]
````

## 🟢 Information: Slices

A slice selects a range of items, using `list[start:stop]`.

The item at `stop` is **not** included.

Let's start at Temple and go three stops.

````python
circle_line_stops[0:3]
````

This returns the items at index `0`, `1`, and `2`.

Remember, slices go up to, but not including, the stop index,
so the stop value (3) is not included.

Then again, if our start value is `0`,
that means the stop value gives us the number of items in the output list.

````python
circle_line_stops[3:6]
````

- If the start value is not given, we take everything from the beginning (index `0`).
- If the stop values is not given, we go right up to the end (index `-1`).

````python
circle_line_stops[:3]
````

````python
circle_line_stops[7:]
````

Bonus:
Note how the output defaults to single quotes except for
"King's Cross St Pancras",
which has `King's` in it and so needs the double.

## 🟢 Information: Numerical functions for lists

Here's a list of approximate annual visitor numbers, in millions, for five London galleries.

````python
gallery_visitors = [5.9, 6.4, 2.1, 2.4, 3.2]
gallery_visitors
````

The command `len` stands for 'length' and gives the number of items in the list.

In this case, that's the number of _galleries_ (not _visitors_).

````python
len(gallery_visitors)
````

`max` and `min` give the 'maximum' (largest) and 'minimum' (smallest) values in the list.

````python
max(gallery_visitors)
````

````python
min(gallery_visitors)
````

`sum` adds up all the values in the list:
in this case this gives the combined number of _visitors_.

````python
sum(gallery_visitors)
````

These can be combined with commands we encountered in Notebook a, such as `round`.

For instance, to calculate the average number of visitors per gallery, rounded to 2 decimal places:

````python
round(sum(gallery_visitors) / len(gallery_visitors), 2)
````

## 🟢 Information: Sorting a list

`sorted` returns a new, sorted list, in ascending order. It does not change the original list.

````python
sorted(gallery_visitors)
````

````python
gallery_visitors  # unchanged
````

Use `reverse=True` to sort in descending order instead.

````python
sorted(gallery_visitors, reverse=True)
````

Lists of text are sorted alphabetically.

````python
sorted(circle_line_stops)
````

The `.sort()` method works differently: it changes the list itself ('in place'), rather than returning a new one.

````python
markets.sort()
markets
````

## 🟢 Information: Dictionaries

A list stores items in order, and finds them by position.

A **dictionary** (short form: `dict`) is another data type.
Dicts store items as `key: value` pairs, in curly brackets `{ }`,
and use the `key` to find the corresponding `value`.

Here are some approximate values for the populations of some London boroughs.

````python
borough_population = {"Camden": 210000, "Islington": 245000, "Hackney": 281000}
borough_population
````

Access a value using its key, in square brackets. I.e.,
- name of dict
- open square brackets `[`
- key
- close square brackets `]`

````python
borough_population["Hackney"]
````

A dictionary does not have numbered indices like a list (see frequent errors below!).

`len` still works, and gives the number of key-value pairs.

````python
len(borough_population)
````

## ⚠️ Common Errors

Here are two common errors to look out for, and how to fix them.

**1. Index out of range**

A list only has as many indices as it has items.

`circle_line_stops` has 27 items (indices 0--26).

Asking for index of `27` (or more) will fail with an `IndexError`.

````python
circle_line_stops[27]
````

Fix:
- check `len(circle_line_stops)` to see how many items there are, and remember the last valid index is one less than the length.
- Alternatively to get the last item regardless of length, use index `-1`.

````python
circle_line_stops[-1]
````

**2. Key not in the dictionary**

Looking up a key that doesn't exist fails, with a `KeyError`.

````python
borough_population["Brent"]
````

Fix:
- check the key is spelled exactly as it appears (capital letters and spacing matter), or check whether it exists first using `in`.
- If it's not there, and you want it to be, then you can add it with the syntax shown below.

````python
"Brent" in borough_population
````

````python
borough_population["Brent"] = 350000
````

````python
"Brent" in borough_population
````

````python
borough_population["Brent"]
````
