## 🟢 Information: Files, folders, and paths: the big picture

Today, phones, tablets, and cloud apps tend to hide how computers organise files,
so many people never need to learn about it.
To work effectively with code and data, you absolutely do need this knowledge.

Let's start with the absolute basics: what is a file, and what is a folder?

**Files and folders**
- A **file** is a named chunk of stored data. Examples include:
    - a photo (yes, it's saved somewhere as a file),
    - this notebook (yes, this is a file too).
- A **folder** (or **directory**) is a container that holds files and other folders.

Folders can contain other folders almost indefinitely.
This structure is often called a **tree**.
Think of a main trunk, then large branches, splitting into smaller branches and so on.
Here's an example based on this resource:

## 🟢 Information: Paths: the address of a file

A **path** is the address of a file. It consists of:
- the sequence of folders you pass through to reach it (in order), and
- the name of the file itself, with its extension.

For example,
`data/Music_venues_grassroots.csv` means "in the folder called `data`, the file called `Music_venues_grassroots.csv`".

There are two kinds of path:
- An **absolute** path starts from the very top of your drive and gives the complete address, e.g.,
    - `/Users/yourname/Documents/python_in_the_smoke/Part_I/data/Music_venues_grassroots.csv` (Mac), or
    - `C:/Users/yourname/Documents/python_in_the_smoke/Part_I/data/Music_venues_grassroots.csv` (Windows).
This works from anywhere on your computer, but it's specific to your computer,
so it won't work on anyone else's.
- A **relative** path starts from *where you are now*, e.g., `data/Music_venues_grassroots.csv`.
This works on any computer, as long as the folders are arranged the same way relative to each other.

This is analogous to the difference between
1. a full postal address (absolute), and
2. directions from where you're standing, like "second on the left" (relative).
Relative directions only make sense if you know where you're starting.

That starting point is called the **working directory**:
the folder Python treats as "here" for this notebook.

Usually it's the folder that contains the notebook itself,
but in some setups it's the folder where you *launched* Jupyter instead.
We can check which with `Path.cwd()`, as below.

## 🟢 Information: You are here!

Python has a package for working with paths, called `pathlib`.
It comes with Python (it's part of the 'standard library'), so we don't need to install it, only import it.

This time, we import one tool from the package rather than the whole package:
`from pathlib import Path` means "from `pathlib`, take the tool called `Path`".

````python
from pathlib import Path
````

Path's command `cwd` stands for the **c**urrent **w**orking **d**irectory.

That is where Python is standing.
Altogether, we call `Path.cwd()`.

````python
Path.cwd()
````

The result is the path of a *folder*, not of a specific file.

You'll see `PosixPath(...)` or `WindowsPath(...)`:
that wrapper simply says what kind of computer the path is for.

The address itself is the text inside the quote marks,
again that should start something like `/Users/yourname/Documents/`.

📝 Note: `Path` appears in two slightly different ways in this notebook.
- `Path.cwd()` uses `Path` *itself* as a tool that answers a question ("where are we?").
- `Path("data")` builds a path **object**, for one specific folder or file.
    - Methods such as `.iterdir()`, `.mkdir()`, `.exists()` and `.resolve()` are then called on that object.
    - Attributes such as `.name` and `.suffix` are read from it.

To see the name of everything in that folder,
we can run a loop (Notebook d) like this:
- `.iterdir()` means "**iter**ate over the **dir**ectory":
it hands us the items in the folder one at a time.
- Each `item` is itself a path,
and `item.name` is its file or folder name (without all the path leading up to that).
Note that there are no brackets after `.name`: it's an attribute, as in Notebook f.
- `sorted` (Notebook b) puts the items in alphabetical order. Without it, `iterdir()` gives them in an arbitrary order.

````python
for item in sorted(Path.cwd().iterdir()):
    print(item.name)
````

## 🟢 Information: Slashes

Folder names in a path are separated by slashes.
- Mac and Linux use a forward slash `/`.
- Windows displays a backslash `\` in File Explorer, but
**Python accepts forward slashes on every system**, including Windows,
so it's simplest always to use `/`.

There's a good reason to avoid the backslash in Python:
the `\` within a string is treated as a special "escape" character (as in `\n` for a new line).
A path like `"C:\Users\name"` will therefore break, or silently change into something else
(see Common Errors below).

📝 Note: some code cells in this section are **supposed** to go wrong (with an error message, or occasionally a surprising result): that's the point!

**1. Wrong path**

Forgetting the folder name, so Python looks in the working directory instead of in `data`: `FileNotFoundError`.

````python
pd.read_csv("Music_venues_grassroots.csv")
````

Fix: compare the path you wrote with the folder tree. Use `Path.cwd()` and `iterdir()` to see where Python is standing and what's there.

Other common reasons for `FileNotFoundError`:
- The file is still in your `Downloads` folder, and was never moved into `data`.
- The extension is hidden and the real file name is something like `Music_venues_grassroots.csv.txt`.
- Your browser renamed the file, for example by adding `(1)` because you downloaded it twice.
- The file name is slightly different in a hard-to-spot way. Here are some classics:
    - Upper/lower case: `Data` vs `data`. (Windows and macOS usually don't mind, but Linux does, and so do most cloud environments.)
    - Look-alike characters: underscore `x_y`, hyphen `x-y`, and space `x y`.

**2. A web page saved instead of the data**

If you save a *web page* instead of the raw CSV file
(for example, with "Save page as" on the dataset's page instead of using the download link),
the file might exist and have the right name, but really be in completely the wrong format.
`read_csv` then usually fails with a `ParserError` (something like `Error tokenizing data`),
or produces a dataframe full of nonsense, because the file is HTML rather than CSV.

Fix: download the *raw* CSV file, as in the exercise.
Verify by opening csvs in a spreadsheet application and/or text editor.
The spreadsheet application should show tabular data;
the text editor should show a file starting with column names
(an HTML file starts with something like `<!DOCTYPE html>`).

**3. Backslashes in a path**

Inside a Python string, a backslash starts what's called an "escape".
For example, you can encode a new line in text with `\n`.
A Windows path copied from File Explorer can therefore break:

````python
"C:\Users\sam\data.csv"
````

This gives a `SyntaxError`: Python reads `\U` as the start of a special character code.

Other combinations don't raise an error at all: they silently change the path.
Here `\n` becomes a *new line*, so the path is split in two:

````python
print("data\new_venues.csv")
````

Some combinations get away with it by luck: `\M` isn't a special code, so Python leaves it alone
(perhaps with a warning).

Arguably that's the worst case: a path that works for one file name and breaks for another.

Fix: use forward slashes, `"data/Music_venues_grassroots.csv"`, on every system.

If you do need to paste a Windows path exactly as it is, put an `r` before the opening quote (a "raw string"),
which switches escapes off: `r"C:\Users\sam\data.csv"`. Forward slashes are still simpler.

**4. The extra `Unnamed: 0` column**

We saw this earlier: `to_csv` without `index=False` writes the index as a column. Fix: add `index=False`.

**5. Saving into a folder that doesn't exist**

`to_csv` creates the file, but not the folder it goes in: `OSError`.

````python
musicians.to_csv("no_such_folder/musicians.csv", index=False)
````

Fix: make the folder first (`Path("no_such_folder").mkdir(exist_ok=True)`), or correct the spelling of the folder name.

**6. `PermissionError` when saving**

On Windows, a file that is still **open in Excel** is locked, so Python can't overwrite it: `PermissionError`.

Fix: close the file in Excel (or save under a different name).

**7. `UnicodeDecodeError` when reading**

Some CSV files (often ones saved by Excel on Windows) are not stored in the usual UTF-8 text encoding.
`read_csv` then fails with a `UnicodeDecodeError`, and the message mentions `utf-8`.

Fix: tell pandas the encoding, e.g. `pd.read_csv("data/file.csv", encoding="cp1252")` (or `"latin-1"`).

**8. Python is not standing where you think**

Your path looks right, but you still get `FileNotFoundError`?
If you launched Jupyter from a different folder, the working directory may not be the folder that holds this notebook.

Fix: run `Path.cwd()`, and compare it with the folder tree.
