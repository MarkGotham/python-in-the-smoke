## 🟢 Information: Working with different types of data

In Python it's easy to tell the difference between numbers and text:
- text is always written with speech marks.
- numbers are never written with speech marks.

For example, this is a number:

```python
42
```

There are two main types of numbers in Python:
- `int` stands for 'integers' (whole numbers)
- `float` is the type which stores numbers with decimal places, e.g., 1.4.

If you're not sure whether something is text or a number, you can use the `type` command to check.

```python
type(42)
```

```python
type(1.4)
```

Below is a piece of text.

```python
"I am learning so much!"
```

In Python, and most programming languages,
text like this is given the 'type' called 'string' (shortened to `str`).

```python
type("I am learning so much!")
```

Does the next cell contains text or a number?

```python
"35"
```

Let's check with the `type` command.

```python
type("35")
```

It's text!

Remember: if something is contained inside speech marks, Python always treats it as text.

## 🟢 Information: Performing calculations

The symbols for addition, subtraction, and division in python are familiar:
- Add: `+`
- Subtract: `-`
- Divide: `/`

```python
3+2
```

```python
2-1
```

```python
4/2
```

📝 Note: although 4/2 is 2, the output is a `float`, not an `int`.

The symbol for multiplication may be more surprising: it is `*`.

```python
4*3
```

As in mathematics, we can use round brackets `()`
to determine the order in which different calculations are made.

The calculations in the innermost brackets are made first.

Note how the commands below give different results.

```python
2*3+5/12
```

```python
((2*3)+5)/12
```

📝 Note: This is potentially confusing. Round brackets now have two different jobs.

We've used them to group a calculation, like `((2*3)+5)/12` above.

We've also used them to call a command, like `print()`.

In fact, both jobs come down to the same idea:
round brackets tell Python "treat what's inside me as one unit."

In the next cell, we demonstrate a combination of both jobs at once. We:
- perform the calculation above (grouping, with round brackets)
- `round` it to a whole number (calling a command, with round brackets)
- return the `type` (necessarily `int`, since we've rounded)

```python
type(round(((2*3)+5)/12))
```

## 🟢 Information: Working with text

As noted above, 'strings' are a way of representing text in Python.

In the Digital Humanities we often need to work with text data, whether that's exhibition catalogues, play scripts, speeches given by public figures, or posts on social media.

We can write strings using single or double speech marks, as long as we're consistent.

```python
print('hello world')
print("hello again!")
```

If in doubt, use double speech marks because you may want to have inverted commas in the text:


This works:

```python
"A Midsummer Night's Dream"
```

This fails:

```python
'A Midsummer Night's Dream'
```

When you run a cell with an error of this kind, you'll see error messages below.

In this case it says that the failure is due to a 'unterminated string literal'.
Broadly speaking, because there are 3 `'` symbols, python can't pair them up and so one is left 'unterminated'.

We'll review this and other common errors near the end of this notebook.

Two strings can be joined together (concatenated) using the **+** operator:

```python
"Hello, " + "world!"
```

Converting the 'case' (uppercase, lowercase, ...) of text works as follows:

```python
'digital humanities'.upper()
```

```python
'DIGITAL HUMANITIES'.lower()
```

```python
'digital humanities'.capitalize()
```

```python
'digital humanities'.title()
```

Note the difference between `capitalize` and `title`.

So far, anything we've done has happened all at once.

That's what the "variable" is for.

The following cells will demonstrate this with the same example immediately above.

```python
"king lear".title()
```

## ⚠️ Common Errors

All programmers make errors.
This is sometimes _frustrating_, but it is certainly _normal_, and it can be a good thing.

One of the joys of programming is that you can
(usually) try something, make a mistake, notice that mistake, and fix it.

In each of these notebooks we'll include a 'frequently made errors' section near the end,
with specific errors relevant to the notebook's topic.

Here's the first batch.

1. We've already encountered the triple (unmatched) single quote mark `'`:

```python
'A Midsummer Night's Dream'
```

This produces an `unterminated string literal` which is a kind of `SyntaxError`.

The error message gives:
- the kind of error: here `SyntaxError`
- more detail `unterminated string literal`
- and the line where it was detected (this will be useful in cells with more lines of code)
- often a carat `^` to indicate where the error took place (or was noted).

If the string contains a single quote, be sure to use double quotes around it:

```python
"A Midsummer Night's Dream"
```

2. Mixing single and double quotes produces the same error type.

Use pairs of single OR double quotes

This fails:

```python
"Midsummer'
```

These two work:

```python
'Midsummer'
```

```python
"Midsummer"
```

3. Mistakes like accidental spaces in a command word produce `SyntaxError`.

```python
p rint("hello world!")
```

Fix: No space!

```python
print("hello world!")
```

4. Missing parentheses is also a `SyntaxError`.

```python
print "hello"
```

5. Trying to 'add' a string to an int gives a `TypeError`.

```python
"hello" + 42
```

6. Trying to call a variable that has not been defined gives a `NameError`.

## ⭐ Optional Exercise: A neater way to combine text and numbers

So far, to mix text and numbers we've had to convert the number to a string first with `str()`.

Python has a shortcut called an **f-string**:
put an `f` right before the opening quote mark,
and put any variable or number inside `{}`.

With this function, no `str()` or `+` are needed.

For example:

```python
exhibition = "Tate Modern"
visitors = 5900000
print(f"{exhibition} welcomes about {visitors} visitors every year.")
```
