## Supporting "Smoke Test" Notebook
---

This supporting "Smoke Test" notebook is simply for making sure installation is in working order.

It's probably mre use for instructors helping to debug than for students.

Simply run is and look for lots of ✅ symbols.

If you see any ❌ symbols, see the guide below.

```python
import sys
from importlib.util import find_spec

print("🐍 Python in the Smoke: smoke test\n")

# 1. Python version
v = sys.version_info
print(f"Python version: {v.major}.{v.minor}.{v.micro}")
if v >= (3, 10):
    print("✅ Version is fine (3.10 or later).")
else:
    print("❌ Too old. You need 3.10 or later. See 'Environment' in the README.")

# 2. Packages (edit this list to match requirements.txt)
for package in ["pandas", "matplotlib"]:
    if find_spec(package):
        print(f"✅ {package} is installed.")
    else:
        print(f"❌ {package} is missing. Run: pip install -r requirements.txt")

# 3. Finish
print("\nIf you can read this, Jupyter is working.")
```

 Symptom                     | 	Likely cause                        | 	Fix                                                    |
-----------------------------|--------------------------------------|---------------------------------------------------------|
 Version ❌	                  | Wrong interpreter or kernel selected | Re-select the `venv` kernel (top right of the notebook)
 Package ❌                   | 	`requirements.txt` not installed	   | Run the `pip install` line in the IDE terminal
 Nothing prints; "no kernel" | Jupyter/ipykernel missing            | 	Let the IDE install it, or `pip install ipykernel`
