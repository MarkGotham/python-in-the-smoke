## *** 🟢 Information: A bar chart for category counts

First summarise, then plot.
Here's the borough count from Notebook i, with the unrecorded borough shown as its own category.
`.plot.barh()` draws a **h**orizontal **bar** chart: a good choice when category names are long.

```python
borough_counts = events["borough"].fillna("Not recorded").value_counts()
borough_counts.plot.barh()
plt.show()
```

That works, but it's hard to read: no title, no axis labels, and the biggest bar is at the bottom.

Let's improve it.
- `.sort_values()` puts the counts in ascending order, so that the largest bar ends up at the top.
- `.plot.barh(...)` returns an **Axes** object: the part of a figure that holds this chart.
We store it as `ax` so that we can add labels.
- `figsize=(8, 4)` gives the width and height in inches, and `color="teal"` sets the colour.
- `ax.set_title(...)`, `ax.set_xlabel(...)` and `ax.set_ylabel(...)` add the labels.
(The `x` axis is the horizontal one, and in a horizontal bar chart that's where the counts are.)
- `ax.set_xticks(range(0, borough_counts.max() + 1))` puts ticks on whole numbers only:
there's no such thing as "half an event" and so our visualisation style should not imply that there is by including sub-integer increments.
- `plt.show()` displays the finished chart.

```python
ax = borough_counts.sort_values().plot.barh(figsize=(8, 4), color="teal")
ax.set_title("Events by recorded borough — fictional programme")
ax.set_xlabel("Number of listed events")
ax.set_ylabel("Borough")
ax.set_xticks(range(0, borough_counts.max() + 1))
plt.show()
```

📝 Notes:
- Bar *lengths* encode counts, so the number axis must start at zero (pandas does this for you; be wary if you ever change the limits).
- The total is 15 events, including one with no recorded borough.
- The bars show this programme's records: they don't measure how much culture a borough has.

Every chart needs the same three labels, so let's write a function for that (Notebook e).
It takes the axes and the three pieces of text, and doesn't need to return anything: it just changes the chart.

```python
def label_chart(ax, title, xlabel, ylabel):
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
```

We'll use `label_chart` from here on.

## *** 🟢 Information: A histogram for a numerical distribution

A histogram groups numbers into intervals called **bins**, and each bar counts how many values fall in that bin.
- The bars count *events*, in intervals of attendance: each bar is a group of events, not a single event.
- We give the bin edges explicitly: `bins=[0, 40, 80, 120, 160, 200]` makes five bins, each 40 wide.
(A value exactly on an edge goes in the bin to its right, except for the very last edge.)
- With only 14 known attendance figures, the shape is very sensitive to the choice of bins.
- Missing attendance is left out: it's not placed in the zero bin.

```python
attendance = events["attendance"].dropna()
ax = attendance.plot.hist(bins=[0, 40, 80, 120, 160, 200], figsize=(8, 4), color="teal", edgecolor="white")
label_chart(ax, f"Recorded attendance — fictional programme (n={len(attendance)})", "Attendees per event", "Number of events")
plt.show()
```

## *** 🟢 Information: A line chart for change through time

A line chart helps us follow the **same measure for the same subject** through time.
We will use daily visits to one exhibition at a **fictional London community gallery**.
Each row describes one calendar day, and every day from 1 to 12 June 2026 is present.
All dates and counts are invented for practice.

This makes a different question possible: **how did daily visits to this exhibition change?**
The event table above mixes different events, capacities and art forms; joining those events
would make changes of event look like changes in a single continuing activity.

Two details matter in this example:
- On 5 June, the visit count was not recorded: the value is unknown.
- On 8 June, the gallery was closed: the recorded count is genuinely zero.

The first few rows are deliberately out of order, so we will need to sort by date.

```python
gallery_visits = pd.DataFrame(
    {
        "date": [
            "2026-06-03", "2026-06-01", "2026-06-02", "2026-06-04",
            "2026-06-05", "2026-06-06", "2026-06-07", "2026-06-08",
            "2026-06-09", "2026-06-10", "2026-06-11", "2026-06-12",
        ],
        "visits": [49, 42, 55, 63, None, 78, 91, 0, 58, 66, 74, 84],
    }
)
gallery_visits["date"] = pd.to_datetime(gallery_visits["date"], format="%Y-%m-%d")
gallery_visits
```

**Predict, then run:** should 5 June appear as a point at zero, a gap, or a point on a line between its neighbours?

- `.sort_values("date")` puts the days in chronological order.
- `.plot.line(x="date", y="visits")` draws the line, and `marker="o"` marks each recorded daily count.
- Keep the row with the unknown count. In this ordinary line plot, a missing value breaks the line.
- Use `.count()` to report how many daily counts are known, and `len(...)` for the number of days in the table.

```python
daily_visits = gallery_visits.sort_values("date")
recorded_days = daily_visits["visits"].count()
print(f"Recorded daily counts: {recorded_days} of {len(daily_visits)}")

ax = daily_visits.plot.line(
    x="date", y="visits", marker="o", figsize=(8, 4), color="teal", legend=False
)
label_chart(
    ax,
    f"Daily exhibition visits — fictional London gallery\n{recorded_days} of {len(daily_visits)} daily counts recorded",
    "Date (June 2026)",
    "Visits per day",
)
ax.set_ylim(bottom=0)
plt.show()
```

There are **11 recorded counts across 12 days**.
The gap at 5 June means "unknown"; the point at zero on 8 June records the closure.
The connecting segments help us follow successive daily counts, but they are not extra observations.
These counts are visits, so someone returning on another day is counted again.

📝 A gap needs a row with a known date and a missing count.
If a day is absent from the table altogether, the plotting command cannot know that an observation was expected:
it will join the surrounding points. Check that every expected day is represented before interpreting the line.

## *** 🟢 Information: Save a chart with its labels

To use a chart in a slide or a document, save it as an image.
- A **figure** is the whole picture; the **axes** are the chart area inside it.
- `ax.get_figure()` gives the figure that contains those axes.
- `fig.savefig(...)` saves that specific figure.
(`plt.savefig(...)` saves whichever figure is "current", which is not always the one you meant.)
- PNG is a good format for slides and documents.
- `dpi=150` sets the resolution, and `bbox_inches="tight"` trims the white space while keeping all the labels.
- Save **before** `plt.show()`.

As in Notebook g, make the folder first, and remember that saving **overwrites** any file with the same name, without asking.
So choose file names deliberately.

```python
Path("figures").mkdir(exist_ok=True)

ax = borough_counts.sort_values().plot.barh(figsize=(8, 4), color="teal")
label_chart(ax, "Events by recorded borough — fictional programme", "Number of listed events", "Borough")
ax.set_xticks(range(0, borough_counts.max() + 1))

fig = ax.get_figure()
fig.savefig("figures/fictional_programme_borough_counts.png", dpi=150, bbox_inches="tight")
plt.show()
```

Check that it's there (Notebook g):

## 🟢 Information: Summary

| To ...                | Use ...                                                              |
|-----------------------|----------------------------------------------------------------------|
| Compare categories    | `counts.sort_values().plot.barh()`                                   |
| Show a distribution   | `numbers.plot.hist(bins=[...])`                                      |
| Compare two numbers   | `df.plot.scatter(x="a", y="b")`                                      |
| See change over time  | `df.sort_values("date").plot.line(x="date", y="b", marker="o")`      |
| Add labels            | `ax.set_title(...)`, `ax.set_xlabel(...)`, `ax.set_ylabel(...)`      |
| Get the figure        | `fig = ax.get_figure()`                                              |
| Save it               | `fig.savefig("figures/chart.png", dpi=150, bbox_inches="tight")`     |
| Show it               | `plt.show()`                                                         |

Always say how many observations are plotted, and why any were left out.
For a daily line chart, keep every expected date and leave unknown counts missing so the line shows a gap.
A recorded zero is still a point; an unknown count is not zero.

## ⚠️ Common Errors

- **`ImportError` or `ModuleNotFoundError` about matplotlib.**
Install it into the notebook's environment (remove the `#` from the `%pip install` cell near the top), as in Notebook f.
- **`KeyError` for a column in a plot.**
Check `events.columns` and the exact spelling.
- **"No numeric data to plot".**
Plot the cleaned number (`price_gbp`), not the original text (`£12`).
- **Plotting an unsummarised category column.**
For a chart of counts, calculate `.value_counts()` first.
- **A line chart that zigzags backwards.**
Line charts join points in the order given: sort by date first.
- **Drawing through a day with no recorded count.**
Dropping its row joins the surrounding points. Keep the dated row with its missing count to show a gap.
If an expected date is absent from the table altogether, represent it with a row whose count is missing.
Do not fill an unknown count with zero.
- **Saving the wrong chart.**
`plt.savefig` saves whichever figure is current.
Use `fig = ax.get_figure()` and `fig.savefig(...)` for the one you mean, and do it before `plt.show()`.
- **`FileNotFoundError` when saving.**
The `figures` folder doesn't exist yet: make it first with `Path("figures").mkdir(exist_ok=True)`.
- **Forgetting the caption.**
Axes describe quantities, but a caption should also say where the data is from, and what was left out.
