## 🟢 Information: Comparing values

Comparing two values in Python always produces a Boolean.

```python
5 > 3
```

```python
5 == 3
```

Here are all the main comparison operators:

| Operator | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Important note:
- the **double** equals sign `==` is for comparing two values.
- A **single** `=` is for assigning a value to a variable (as we've seen in Notebooks a and b).
- Mixing these up is one of the most common early mistakes. See the Common Errors section below.

```python
borough_population = {"Camden": 210000, "Islington": 245000, "Hackney": 281000}
borough_population["Islington"] > borough_population["Camden"]
```

## 🟢 Information: Combining comparisons with `and` / `or`

Sometimes more than one condition needs to be checked at once.

- `and`: both conditions must be `True`
- `or`: at least one condition must be `True`

```python
venue_prices = {"Tate Modern": 0, "London Eye": 29, "Shakespeare's Globe": 25}
venue_prices["Tate Modern"] == 0 and venue_prices["London Eye"] == 0
```

```python
venue_prices["Tate Modern"] == 0 or venue_prices["London Eye"] == 0
```

## 🟢 Information: `if` and `else`

A Boolean can control which code runs, using `if`.

```python
if venue_prices["Tate Modern"] == 0:
    print("Tate Modern is free to enter.")
```

Two things to notice:
- The `if` line ends with a colon `:`
- The line below is **indented** (usually four spaces). Python uses this indentation to know which code belongs to the `if`.

Add `else` to run different code when the condition is `False`.

```python
if venue_prices["London Eye"] == 0:
    print("The London Eye is free to enter.")
else:
    print("The London Eye is not free to enter.")
```

## 🟢 Information: `elif`

Use `elif` (short for 'else if') to check several conditions in order.

After an `if`, you can use as many instances of `elif` as you like.

To be clear:
- One `if` is required
- One or more `elif` instances can follow
- There may or may not be one single `else` at the end (no more than 1).

Python stops at the first one that is `True`; the final `else` catches everything else.

Let's add an expensive museum trip to our dict:

```python
venue_prices["British Museum London Guided Tour"] = 90
```

Now let's retrieve that value and check it against if/elif/elif/else options:

```python
price = venue_prices["British Museum London Guided Tour"]

if price < 0:  # Some mistake surely?
    print("Wait, you are paying me to go to a museum?")
elif price == 0:  # This scenario only: exactly 0
    print("Free")
elif (
    price < 20
):  # Less than 20 (but given the above conditions, we're also testing greater than 0
    print("Budget")
elif price < 40:  # Likewise, less than 40 but greater than or equal to 20.
    print("Standard")
else:  # Any other value: here that means more than 40
    print("Premium!")
```

## ⚠️ Common Errors

Three errors to look out for, and how to fix them.

**1. Using `=` instead of `==`**

A single `=` assigns a value; a double `==` compares two values.

Using the wrong one in a condition causes a `SyntaxError`.

```python
if price = 0:
    print("Free")
```

Fix: use `==` to compare.

```python
if price == 0:
    print("Free")
```

**2. Missing colon**

Every `if`, `elif`, and `else` line must end with a colon `:`.

Leaving it off causes a `SyntaxError`.

```python
if price == 0
    print("Free")
```

Fix: add the colon.

```python
if price == 0:
    print("Free")
```

**3. Missing or inconsistent indentation**

The line(s) below `if`, `elif`, or `else` must be indented.
(usually four spaces, Jupyter and most other editors add this automatically after a colon).

Missing indentation causes an `IndentationError`.

```python
if price == 0:
print("Free")
```

Fix: indent the line below the colon.

```python
if price == 0:
    print("Free")
```
