## 🟢 Information: The `for` loop

A `for` loop repeats a block of code once for every item in a list.

````
tube_lines = [
    "Bakerloo",
    "Central",
    "Circle",
    "District",
    "Jubilee",
    "Metropolitan",
    "Northern",
    "Piccadilly",
    "Victoria",
    "Waterloo & City",
]

for line in tube_lines:
    print(line)
````

As with `if` in Notebook c:
- The `for` line ends with a colon `:`
- The line(s) below are indented

`line` here is a new variable, created by the loop. It takes the value of each item in `tube_lines`, one at a time, until the list runs out.

## 🟢 Information: Looping with `range`

Sometimes we want to repeat an action a fixed number of times, without looping over a list.

For that kind of task, use `range`.

`range` generates a sequence of numbers to loop over.

````
for number in range(5):
    print(number)
````

📝 Important note:
- `range(5)` produces `0, 1, 2, 3, 4`
- as before, this starts at `0` (not `1`)
- as before, this ends at `4` (does not include 5).

This is the same 'start counting from zero' rule as list indices in Notebook b.

## 🟢 Information: Building up a total (the 'accumulator' pattern)

For loops can be used to build up a result across every item in a list.

````
park_sizes = [
    142,
    166,
    111,
    253,
    32,
]  # approximate size in hectares of five London parks

total = 0

for size in park_sizes:
    total = total + size

total
````

What's happening here:
- `total` starts at `0`, before the loop begins
- Each time round the loop, the current park's size is added to `total`
- After the loop finishes, `total` holds the sum of every park size

Isn't this the same result that we achieved with the built-in `sum` function in Notebook b?
Yes, well spotted!
Sometimes it's useful to write it out with a loop.
Moreover, loops can be used for any calculation, beyond simply `sum`.

## 🟢 Information: Looping over a dictionary

Looping over a dictionary directly (as above) gives the keys.

To get both the key and the value together, use `.items()`.

````
for musician, borough in musician_boroughs.items():
    print(musician, "was born in", borough)
````

`.items()` gives each key-value pair, one at a time.

`musician` and `borough` are just variable names.
Python assigns the key to the first and the value to the second, in order.

## 🟢 Information: Combining a loop with a conditional

A loop can apply an `if` / `elif` / `else` from Notebook c to every item.

This saves us from checking each single value on its own.

````
venue_prices = {"Tate Modern": 0, "London Eye": 29, "Shakespeare's Globe": 25}

for venue, price in venue_prices.items():
    if price == 0:
        print(venue, "is Free")
    elif price < 20:
        print(venue, "is Budget")
    elif price < 40:
        print(venue, "is Standard")
    else:
        print(venue, "is Premium")
````

## ⚠️ Common Errors

Two errors to look out for, and how to fix them.

**1. Using the wrong variable name inside the loop**

The loop variable (e.g. `line`, `musician`) only exists under the name given after `for`.

Using a different name inside the loop causes a `NameError`.

````
for musician in musician_boroughs:
    print(name)
````

Fix: use the same name that appears after `for`.

````
for musician in musician_boroughs:
    print(musician)
````

**2. Index out of range with `range(len(...))`**

Looping with `range(len(tube_lines))` gives valid indices `0` to `9` for a 10-item list.

Writing `range(len(tube_lines) + 1)`, or forgetting that `range` excludes its endpoint,
produces one index too many, and an `IndexError`.

````
for i in range(len(tube_lines) + 1):
    print(tube_lines[i])
````

Fix:
- minimally: use `range(len(tube_lines))`, without the `+ 1`
- better: loop over the list directly, as at the top of this notebook, which avoids indices altogether.

````
for i in range(len(tube_lines)):
    print(tube_lines[i])
````

## ⭐ Optional Exercise: A neater way to loop with a position

`range(len(...))` works, but Python has a neater tool for looping over a list while also keeping track of each item's position: `enumerate`.

````
for position, line in enumerate(tube_lines):
    print(position, line)
````

`enumerate` gives both the index and the item together, without needing `range`, `len`, or indexing with `[ ]`.

Optional: use `enumerate` to print each London park's position.

(And remember that, by default, this starts from 0, so if you want to count from 1, then `+1` to each).
